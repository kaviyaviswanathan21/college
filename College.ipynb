# How many standard deviations away from the national college average SAT score are Stanford's and MIT's average SAT scores?

To find the **z-scores** for the average SAT scores of Stanford and MIT, I will be conducting an **observational study** because I am analyzing data that already exists across the colleges in America without committing any changes to them. I will use the overall mean and standard deviation to calculate their z-scores, allowing me to compare which university has a higher standardized SAT average compared to all other colleges in the dataset.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

college = pd.read_csv("cleaned_college.csv")
college.head()
```

### Overview: 

```python
college.info()
```

### Average SAT Score at Stanford:

```python
stanford_score = college[college['school_name'] == "Stanford University"]["sat_average"].item()
print(stanford_score)
```

### Average SAT Score at MIT:

```python
mit_score = college[college['school_name'] == "Massachusetts Institute of Technology"]["sat_average"].item()
print(mit_score)
```

### Average Mean and Standard Deviation of National Colleges in America:

```python
mean_sat = college["sat_average"].mean()
print("Mean SAT score: " , mean_sat)
std_sat = college["sat_average"].std()
print(std_sat)
```

### Z Score Evaluation: 

```python
mit_z = (mit_score - mean_sat) / std_sat
stanford_z = (stanford_score - mean_sat) / std_sat

print(f"MIT SAT Z-Score: {mit_z:.2f}")
print(f"Stanford SAT Z-Score: {stanford_z:.2f}")
```

## Visuals:

### Histogram:

```python
plt.figure(figsize=(8, 5))
sns.histplot(college["sat_average"], bins=20, kde=True, color="skyblue")
plt.axvline(mean_sat, color="red", linestyle="--", label=f"Mean: {mean_sat:.0f}")
plt.axvline(stanford_score, color="green", linestyle="--", label="Stanford")
plt.axvline(mit_score, color="purple", linestyle="--", label="MIT")
plt.title("SAT Average Distribution")
plt.xlabel("SAT Average")
plt.ylabel("Frequency")
plt.legend()
plt.tight_layout()
plt.show()
```

### Box Plot:

```python
plt.figure(figsize=(6, 4))
sns.boxplot(x=college["sat_average"], color="pink")
plt.title("SAT Average Box Plot")
plt.xlabel("SAT Average")
plt.tight_layout()
plt.show()
```

# Conclusion:

After evaluating this data, I have been able to compare Stanford and MIT's SAT scores to every other national college in America. Stanford's SAT Z-score came out to be 2.48 standard deviations away from the average and MIT's SAT Z-score came out to be 2.53 standard deviations away from the average. Both of these z-scores would be considered unusually high and this can be seen in the histogram where the average SAT scores for Stanford and MIT are both further right relative to the average SAT scores across America. After looking at this data, we can conclude that Stanford and MIT are better preforming colleges compared to the other colleges in America.
